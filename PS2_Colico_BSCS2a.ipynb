#### Problem 2: Image Thresholding

```python

import cv2
import numpy as np 
import math
im_file ="1.jpg" #Read the image file
tru_img = cv2.imread(im_file,0)#read the image file 
img = cv2.imread(im_file, 0)
threshold = 80# selcted threshold

# get image properties.
h, w = np.shape(img)# properties include is height and weight of the selected image
##### Applying binary thresholding to an image.
# iterate over the entire image and convert to Grayscale.
for py in range(0, h):
    for px in range(0, w):
        if img[py][px] >= threshold:# the threshold is iterate to the entire image
            img[py][px]=255# 255 the maximum pixel value is selected
        elif img[py][px] <= threshold:
            img[py][px]=0

cv2.imwrite('problem2output.jpg', img)
cv2.imshow('True', tru_img)
cv2.imshow('inverted', img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### Problem 3: Logarithmic Transformation

```python
import cv2
import numpy as np 
import math #math is import for the logic of the logarithmics function dependencies
```

###### Calling the Image File

```python
im_file ="1.jpg" # the file image 
true_img = cv2.imread(im_file, 0)
img = cv2.imread(im_file, 0)
constant = 50 # constant output specify to the img
```

#### Applying logarithmic transformation to an image

```python
h, w = np.shape(img)
img = img.astype(np.float32)  # Convert to float
# iterate over the entire image and convert to Grayscale.
for py in range(0, h):
    for px in range(0, w):
        img[py][px] = constant * math.log(1 + img[py][px])
```

###### Write and Read the Input and Output Function

```python
cv2.imwrite('problem3_out.jpg', img)
cv2.imshow('original', true_img)# Display original image
cv2.imshow('inverted', img)# Display inverted version of the  image

cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### Problem 4: Power-Law (Gamma) Transformation

Importing Important Libraries Dependencies

```python
import cv2
import numpy as np 
import math
```

Applying Gamma Input variables to the Image

```python
im_file ="1.jpg" # the file image 
true_img = cv2.imread(im_file, 0)# read the original img file
img = cv2.imread(im_file, 0)
#The define intensity for gamma transformation
constant = 40
gamma = 0.4
```

```python
h, w = np.shape(img)## Get image dimensions (height and width)
# iterate over the entire image and convert to Grayscale.
for py in range(0, h):
    for px in range(0, w):
        img[py][px] = constant * (img[py][px] ** gamma)
```

```python
# Write and Read the Input and Output Function
cv2.imwrite('problem4_out.jpg', img)
cv2.imshow('original', true_img)
cv2.imshow('inverted', img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### Problem 5: Gray Level Slicing

Importing Important Dependencies

```python
import cv2
import numpy as np 
```

Loading and Retrieving image I min nad I max

```python
im_file ="1.jpg"
true_img = cv2.imread(im_file, 0)# Read image in grayscale
img1 = cv2.imread(im_file, 0)
img2 = cv2.imread(im_file, 0)

# The define intensity for gray level slicing
I_min = 100
I_max = 200
```

Function structure to perform gray-level slicing in two modes

```python
h, w = np.shape(true_img)# Get image dimensions (height and width)
for py in range(0, h):
    for px in range(0, w):
        if I_min <= img1[py][px] <= I_max: # Applying gray-level slicing without background preservation
            img1[py][px] = 255 # max (255)
        else:
            img1[py][px] = 0 #min (0)
        if not (I_min <= img2[py][px] <= I_max):
            img2[py][px] = true_img[py][px] 
```

###### Write and Read the Input and Output Function

```python
cv2.imwrite('Problem5.1output.jpg', img1)# Image without background preservation
cv2.imwrite('Problem5.2output.jpg', img2) # Image with background preservation
# Extract and display each gray level
cv2.imshow('Original', true_img)
cv2.imshow('Without background preservation', img1)
cv2.imshow('With background preservation', img2)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

##### Problem 6: Bit-Plane Slicing

Importing Important Libraries Dependencies

```python
import cv2
import numpy as np
```

```python
def extract_bit_planes(image_path):
    # Load grayscale image
    true_img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    
    if true_img is None:
        raise ValueError("Image not found or unable to load.")

    h, w = true_img.shape  # Get image dimensions
    bit_planes = []  # List to store extracted bit planes

    # Extract and display each bit plane
    for bit in range(8):
        bit_plane = (true_img >> bit) & 1  # Extract bit-plane
        bit_plane_img = bit_plane * 255  # Scale for visualization
        bit_planes.append(bit_plane_img.astype(np.uint8))

        # Save bit-plane image
        cv2.imwrite(f'Problem 6_bit_planes_{bit}.jpg', bit_plane_img)
        cv2.imshow(f'Problem 6_Bit Planes {bit}', bit_plane_img)

    # Display original image
    cv2.imshow('Original Image', true_img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

# Example usage
extract_bit_planes("1.jpg")
```
